```python
batsman = pd.read_csv('/content/sharma-kohli.csv')
batsman
```

```python
plt.plot(batsman['index'],batsman['V Kohli'])
```

```python
plt.plot(batsman['index'],batsman['V Kohli'])
plt.plot(batsman['index'],batsman['RG Sharma'])
```

```python
plt.plot(batsman['index'],batsman['V Kohli'])
plt.plot(batsman['index'],batsman['RG Sharma'])
plt.title('Rohit Sharma Vs. Virat Kohli Runs In Every Season')
plt.xlabel('Season')
plt.ylabel('Run scored')
```

```python
plt.plot(batsman['index'],batsman['V Kohli'],color='green',linestyle='solid',linewidth=5,marker='D')
plt.plot(batsman['index'],batsman['RG Sharma'],color='black',linestyle='dotted',linewidth=2,marker='*',markersize=10)
plt.title('Rohit Sharma Vs. Virat Kohli Runs In Every Season')
plt.xlabel('Season')
plt.ylabel('Run scored')
```

```python
plt.plot(batsman['index'],batsman['V Kohli'],color='green',linestyle='solid',linewidth=5,marker='D',label='Virat')
plt.plot(batsman['index'],batsman['RG Sharma'],color='black',linestyle='dotted',linewidth=2,marker='*',markersize=10,label='Rohit sharma')
plt.title('Rohit Sharma Vs. Virat Kohli Runs In Every Season')
plt.xlabel('Season')
plt.ylabel('Run scored')
plt.legend(loc='best')
plt.grid()
```

```python
plt.plot(batsman['index'],batsman['V Kohli'],color='green',linestyle='solid',linewidth=5,marker='D',label='Virat')
plt.plot(batsman['index'],batsman['RG Sharma'],color='black',linestyle='dotted',linewidth=2,marker='*',markersize=10,label='Rohit sharma')
plt.title('Rohit Sharma Vs. Virat Kohli Runs In Every Season')
plt.xlabel('Season')
plt.ylabel('Run scored')
plt.legend(loc='upper right')
```
