```python
### Q2:- Write a program that will convert celsius value to fahrenheit.

#take celsius as input from the user 
celsius = int(input("Enter celsius : "))
#make a variable as F to store fahrenheit 
Fahrenheit = celsius * (9/5) + 32
#print fahrenheit as F
print("Fahrenheit for given ",celsius," would be : ",Fahrenheit," fahrenheit.")
```

```python
### Q4:- Write a program to find the euclidean distance between two coordinates.Take both the coordinates from the user as input.

x1 = int(input("Enter x1 : "))
y1 = int(input("Enter y1 : "))
x2 = int(input("Enter x2 : "))
y2 = int(input("Enter y2 : "))
result = ((x2 - x1)**2 + (y2 - y1)**2) ** 0.5
print(result)
```

```python
### Q5:- Write a program to find the simple interest when the value of principle,rate of interest and time period is provided by the user.

p = int(input("Enter the principle : "))
r = int(input("Enter rate of interest : "))
t = float(input("Enter time period in year: "))

SI = (p*r*t)/100
print(SI)
```

```python
### Q6:- Write a program that will tell the number of dogs and chicken are there when the user will provide the value of total heads and legs.

#For example:
#Input:
#heads -> 4
#legs -> 12
#Output:
#dogs -> 2
#chicken -> 2

#take input from the user 

total_heads = int(input("Enter no. of heads : "))
total_legs = int(input("Enter no. of legs : "))

d = (total_legs - 2 * total_heads) // 2
c = total_heads - d 

#print('c : ',c)
#print('d : ',d)

if d >= 0 and c >= 0 and total_legs % 2 == 0:
    print("Dogs:", d)
    print("Chickens:", c)
else:
    print("Invalid input! The given number of heads and legs do not match real animals.")



```

```python
### Q7:- Write a program to find the sum of squares of first n natural numbers where n will be provided by the user.

n = int(input("Enter the value : "))

sum_sqr = n*(n+1)*(2*n+1)//6

print(sum_sqr)
```

```python
### Q8:- Given the first 2 terms of an Arithmetic Series.Find the Nth term of the series. Assume all inputs are provided by the user.


first_term = int(input("Enter first term : "))
second_term = int(input("Enter the second term : "))
n = int(input("Enter nth term : "))

d = second_term - first_term 

an = first_term + (n-1)*d

print(an)
```
